# 0. Definiciones y librería para el análisis.

````r
# @title Función para agrupar y resumir los datos por intervalos
#'
#' @param ancho Numérico. El ancho de los tramos o intervalos.
#' @param set Data frame. El conjunto de datos de entrada.
#' @param resumir Texto. Nombre de la variable numérica a agrupar (por ejemplo, 'x1').
#' @param agrupacion Texto. Columnas para separar los grupos (por ejemplo, 'tratamiento').
resumen <- function(
        ancho,
        set,
        resumir,
        agrupacion = NULL)
    {
    variable_a_resumir = resumir
    factores_agrupacion = agrupacion
    # Obtener los valores de la variable a resumir de forma dinámica
    valores <- set[[variable_a_resumir]]

    # Definir los cortes del intervalo en base al ancho solicitado
    cortes <- seq(
        floor(min(valores, na.rm = TRUE) / ancho) * ancho,
        ceiling(max(valores, na.rm = TRUE) / ancho) * ancho,
        by = ancho
        )

    # Asignar cada fila a su tramo correspondiente
    datos_tramos <- set
    datos_tramos$tramo <- cut(valores, breaks = cortes, include.lowest = TRUE)
    datos_tramos$n <- 1L

    # Construir la parte derecha de la fórmula dinámicamente según los factores provistos
    variables_grupo <- c("tramo", factores_agrupacion)
    formula_lado_derecho <- paste(variables_grupo, collapse = " + ")
    formula_completa <- as.formula(
        paste("cbind(n, fracturas = ever, fragilidad =", variable_a_resumir, ") ~", formula_lado_derecho)
        )

    # Agrupar y sumar los datos
    resumen_df <- aggregate(
        formula_completa,
        data = datos_tramos,
        FUN = sum,
        na.action = na.fail
        )

    # Calcular la posición central promedio y la tasa de eventos del tramo
    resumen_df$fragilidad_u <- resumen_df$fragilidad / resumen_df$n
    resumen_df$fracturas_u <- resumen_df$fracturas / resumen_df$n

    # Calcular las odds de fracturas_u: p / (1 - p)
    resumen_df$odds_fracturas <- resumen_df$fracturas_u / (1 - resumen_df$fracturas_u)

    return(resumen_df)
    }
````

````r
# @title Función de visualizar resumen y modelo
visualizar_ajuste <- function(datos_resumen, modelo) {

  # 1. Preparar los puntos observados
  etiquetas <- c("Sin tratamiento", "Con tratamiento")
  tiene_tratamiento <- "tratamiento" %in% names(datos_resumen)
  por_tratamiento <- "x2" %in% all.vars(formula(modelo))

  puntos <- transform(datos_resumen, x1 = fragilidad_u)

  if (tiene_tratamiento) {
    puntos$x2 <- match(puntos$tratamiento, etiquetas) - 1L

    if (anyNA(puntos$x2)) {
      stop("Tratamiento no reconocido.")
    }
  } else {
    if (por_tratamiento) {
      stop("El modelo incluye x2: utiliza un resumen por tratamiento.")
    }

    puntos$tratamiento <- "Global"
  }

  puntos$prediccion <- predict(modelo, newdata = puntos, type = "response")
  puntos$error_abs <- abs(puntos$fracturas_u - puntos$prediccion)

  # 2. Obtener las líneas del modelo
  lineas <- expand.grid(
    x1 = seq(min(puntos$x1), max(puntos$x1), length.out = 100),
    x2 = if (por_tratamiento) unique(puntos$x2) else 0L
  )

  lineas$tratamiento <- if (por_tratamiento) {
    etiquetas[lineas$x2 + 1L]
  } else {
    "Modelo"
  }

  lineas$prediccion <- predict(modelo, newdata = lineas, type = "response")

  # 3. Comparar observaciones y predicciones
  ggplot(puntos, aes(x1, fracturas_u, colour = tratamiento)) +
    geom_line(
      data = lineas, aes(y = prediccion),
      linewidth = 1.2, show.legend = FALSE
    ) +
    geom_point(
      aes(size = n, fill = error_abs),
      shape = 21, stroke = 1.5, alpha = 0.8
    ) +
    geom_vline(
      xintercept = mean(puntos$x1),
      linetype = "dotted", colour = "red", linewidth = 0.8
    ) +
    geom_hline(
      yintercept = mean(puntos$fracturas_u),
      linetype = "dotted", colour = "red", linewidth = 0.8
    ) +
    scale_colour_manual(
      values = c(
        "Sin tratamiento" = "Navy",
        "Con tratamiento" = "DarkGreen",
        "Global" = "grey35",
        "Modelo" = "grey35"
      ),
      breaks = etiquetas,
      guide = if (tiene_tratamiento) "legend" else "none"
    ) +
    scale_fill_gradient(low = "goldenrod", high = "darkmagenta") +
    labs(
      title = "OsteoRed: proporciones observadas y modelo",
      subtitle = paste(class(modelo)[1], "—", deparse(formula(modelo))),
      x = "Fragilidad ósea estandarizada",
      y = "Proporción observada / predicción",
      colour = "Tratamiento",
      fill = "Discrepancia absoluta\npor tramo",
      size = "Pacientes (n)",
      caption = "Referencias rojas: medias no ponderadas de los puntos agrupados."
    ) +
    theme_minimal(base_size = 12)
}
````

````r
# @title Diagnósticar Modelo
diagnosticar <- function(modelo) {
    es_glm <- inherits(modelo, "glm")

    if (!inherits(modelo, "lm")) {
    stop("Usa un modelo ajustado con lm() o glm().")
    }

    if (es_glm) {
    if (family(modelo)$family != "binomial" ||
        !all(modelo$prior.weights == 1)) {
        stop("Esta versión admite GLM binomiales individuales sin ponderaciones.")
    }
    } else if (!is.null(modelo$weights)) {
    stop("Esta versión admite modelos lm() sin ponderaciones.")
    }

    observado <- model.response(model.frame(modelo))

    if (!is.numeric(observado) || !is.null(dim(observado)) ||
        !all(observado %in% c(0, 1))) {
        stop("La respuesta debe ser un vector numérico codificado como 0/1.")
    }

    datos <- data.frame(
        ajustado = modelo$fitted.values,
        residuo = observado - modelo$fitted.values,
        respuesta = factor(observado, levels = c(0, 1), labels = c("No", "Sí"))
    )

    tipo <- if (es_glm) {
        paste0("GLM binomial (", family(modelo)$link, ")")
        } else {
        "OLS"
        }

        subtitulo <- paste(
        tipo,
        paste(deparse(formula(modelo)), collapse = " "),
        sep = ": "
        )

    g_residuos <- ggplot(datos, aes(ajustado, residuo)) +
        geom_hline(yintercept = 0, linetype = "dashed", colour = "grey45") +
        geom_vline(xintercept = c(0, 1), linetype = "dotted", colour = "grey65") +
        geom_point(
        aes(fill = respuesta),
        shape = 21, colour = "grey35",
        size = 1.8, stroke = 0.3, alpha = 0.65
        ) +
        scale_fill_manual(values = c("No" = "Tomato", "Sí" = "Gold")) +
        labs(
        title = "Residuos frente a ajustados",
        subtitle = subtitulo,
        x = "Valor ajustado",
        y = "Residuo: observado − ajustado",
        fill = "Respuesta",
        caption = "Las verticales señalan los límites 0 y 1."
        ) +
        theme_minimal(base_size = 12) +
        theme(legend.position = "bottom")

    # Para el GLM binomial mostramos los residuos de respuesta.
    if (es_glm) {
        return(g_residuos)
        }
    g_qq <- ggplot(datos, aes(sample = residuo)) +
        stat_qq(
        shape = 21, fill = "grey75", colour = "grey35",
        size = 1.4, alpha = 0.65
        ) +
        stat_qq_line(colour = "firebrick", linewidth = 0.7) +
        labs(
        title = "QQ normal de los residuos",
        subtitle = subtitulo,
        x = "Cuantiles teóricos normales",
        y = "Cuantiles de los residuos"
        ) +
        theme_minimal(base_size = 12)

    patchwork::wrap_plots(g_residuos, g_qq, ncol = 2)
    }
````

````r
# @title Resumen y modelo en cuatro sistemas de coordenadas
a_plot <- function(datos_resumen, modelo) {

  # 1. Comprobar el modelo y las columnas del resumen
  if (!inherits(modelo, "glm")) {
    stop("Usa el GLM binomial, por ejemplo modelo_fractura.")
  }

  familia <- family(modelo)

  if (!familia$family %in% c("binomial", "quasibinomial") ||
      !familia$link %in% c("logit", "probit", "cloglog", "cauchit")) {
    stop("Se necesita un enlace biyectivo entre (0, 1) y R.")
  }

  campos <- c("fragilidad_u", "fracturas_u", "n")

  if (!all(campos %in% names(datos_resumen))) {
    stop("El resumen debe contener fragilidad_u, fracturas_u y n.")
  }

  if (!nrow(datos_resumen) ||
      !all(vapply(datos_resumen[campos], is.numeric, logical(1))) ||
      any(!is.finite(as.matrix(datos_resumen[campos]))) ||
      any(datos_resumen$fracturas_u < 0 |
          datos_resumen$fracturas_u > 1) ||
      any(datos_resumen$n <= 0)) {
    stop("Revisa los valores del resumen.")
  }

  α <- familia$linkinv
  g <- familia$linkfun

  etiquetas <- c("Sin tratamiento", "Con tratamiento")
  por_tratamiento <- "x2" %in% all.vars(formula(modelo))
  tiene_tratamiento <- "tratamiento" %in% names(datos_resumen)

  # 2. Preparar las bolas: proporciones observadas por tramo
  puntos <- transform(
    datos_resumen,
    x1 = fragilidad_u,
    p = fracturas_u
  )

  if (tiene_tratamiento) {
    puntos$x2 <- match(puntos$tratamiento, etiquetas) - 1L

    if (anyNA(puntos$x2)) {
      stop("Tratamiento no reconocido.")
    }
  } else {
    if (por_tratamiento) {
      stop("El modelo incluye x2: usa resumen_tratamiento.")
    }

    puntos$tratamiento <- "Global"
  }

  puntos$η <- g(puntos$p)

  puntos$error_abs <- abs(
    puntos$p - predict(modelo, puntos, type = "response")
  )

  # 3. Preparar las curvas en las coordenadas originales
  lineas <- expand.grid(
    x1 = seq(min(puntos$x1), max(puntos$x1), length.out = 200),
    x2 = if (por_tratamiento) sort(unique(puntos$x2)) else 0L
  )

  lineas$tratamiento <- if (por_tratamiento) {
    etiquetas[lineas$x2 + 1L]
  } else {
    "Modelo"
  }

  lineas$η <- predict(modelo, lineas, type = "link")
  lineas$p <- predict(modelo, lineas, type = "response")

  # 4. Aplicar las mismas transformaciones a cualquier capa
  cuatro_vistas <- function(datos) {
    vistas <- rbind(
      transform(datos, X = x1,    Y = η, escala_x = 1L, escala_y = 1L),
      transform(datos, X = α(x1), Y = η, escala_x = 2L, escala_y = 1L),
      transform(datos, X = x1,    Y = p, escala_x = 1L, escala_y = 2L),
      transform(datos, X = α(x1), Y = p, escala_x = 2L, escala_y = 2L)
    )

    vistas$escala_x <- factor(
      vistas$escala_x, 1:2,
      c("bold(R) ~ ':' ~ x[1]",
        "alpha(bold(R)) ~ ':' ~ alpha(x[1])")
    )

    vistas$escala_y <- factor(
      vistas$escala_y, 1:2,
      c("bold(R) ~ ':' ~ g(p)",
        "alpha(bold(R)) ~ ':' ~ p")
    )

    vistas
  }

  referencias <- data.frame(
    x1 = mean(puntos$x1),
    p = mean(puntos$p)
  )
  referencias$η <- g(referencias$p)

  referencias <- cuatro_vistas(referencias)
  puntos <- cuatro_vistas(puntos)
  lineas <- cuatro_vistas(lineas)

  # Los infinitos se indican con triángulos en los bordes
  puntos$forma <- ifelse(
    is.infinite(puntos$Y),
    ifelse(puntos$Y < 0, 25, 24),
    21
  )

  # 5. Dibujar las cuatro vistas con leyendas compartidas
  ggplot() +
    geom_vline(
      data = referencias, aes(xintercept = X),
      linetype = "dotted", colour = "red", linewidth = 0.5
    ) +
    geom_hline(
      data = referencias, aes(yintercept = Y),
      linetype = "dotted", colour = "red", linewidth = 0.5
    ) +
    geom_line(
      data = lineas,
      aes(X, Y, colour = tratamiento),
      linewidth = 0.9, show.legend = FALSE
    ) +
    geom_point(
      data = puntos,
      aes(
        X, Y,
        colour = tratamiento,
        fill = error_abs,
        size = n,
        shape = forma
      ),
      stroke = 1.1, alpha = 0.85
    ) +
    facet_grid(
      escala_y ~ escala_x,
      scales = "free",
      labeller = label_parsed
    ) +
    scale_shape_identity() +
    scale_size_area(max_size = 5) +
    scale_colour_manual(
      values = c(
        "Sin tratamiento" = "Navy",
        "Con tratamiento" = "DarkGreen",
        "Global" = "grey35",
        "Modelo" = "grey35"
      ),
      breaks = etiquetas,
      guide = if (tiene_tratamiento) "legend" else "none"
    ) +
    scale_fill_gradient(
      low = "goldenrod",
      high = "darkmagenta"
    ) +
    coord_cartesian(clip = "off") +
    labs(
      title = "OsteoRed: el mismo resumen y modelo en cuatro coordenadas",
      subtitle = paste(
        paste(deparse(formula(modelo)), collapse = " "),
        "· enlace", familia$link
      ),
      x = "Fragilidad: coordenada indicada en cada columna",
      y = "Respuesta: coordenada indicada en cada fila",
      colour = "Tratamiento",
      fill = "|p observada − p modelo|",
      size = "Pacientes (n)",
      caption = paste(
        "Triángulos en los bordes: g(0) = −∞ y g(1) = +∞; no son valores finitos.",
        "Referencias rojas: transformación de las mismas medias de tramos (sin ponderar).",
        sep = "\n"
      )
    ) +
    theme_minimal(base_size = 11) +
    theme(
      legend.position = "bottom",
      legend.box = "vertical",
      panel.spacing = grid::unit(1.4, "lines")
    )
}
````

#  Modelar una probabilidad binomial
Podemos expresar una misma probabilidad mediante distintas coordenadas:
$$
p
\quad\longleftrightarrow\quad
\frac{p}{1-p}
\quad\longleftrightarrow\quad
\log\frac{p}{1-p}.
$$
Cada representación facilita una operación diferente:
- Probabilidad: permite interpretar el resultado.
- Odds: permiten comparar el evento con su contrario.
- Logit: permite expresar cambios multiplicativos de las odds como sumas.

Convertir una probabilidad en logit es una operación algebraica. Proponer que ese logit depende linealmente de fragilidad y tratamiento es una hipótesis del modelo.

#🔰 Bernoulli: ¿qué estamos intentando modelar?
En la cohorte, una paciente $i$ tiene un resultado observado:
$$
y_{[i]}\in\{0,1\}.
$$
Pero antes de observarlo podemos asignar una probabilidad a cada posibilidad:
$$
P(Y_{[i]}=1)=p_{[i]},
\qquad
P(Y_{[i]}=0)=1-p_{[i]}.
$$

Eso es una variable Bernoulli.

Para nuestro modelo, esas probabilidades están condicionadas por la fragilidad y el tratamiento.

la media de una respuesta Bernoulli es su probabilidad de valer 1.
$$
E[Y]=0(1-p)+1p=p.
$$
Y su varianza también se deduce directamente.

$$
\mathrm{Var}(Y)
=E[Y^2]-E[Y]^2
=p-p^2
=p(1-p).
$$

Esta relación explica por qué la variación cambia con la probabilidad: cerca de 0 o 1 hay menos incertidumbre que cerca de 0.5. Bernoulli es precisamente el caso de una binomial con un ensayo.

Las odds comparan fractura con ausencia de fractura:
$$
\frac{p}{1-p}
=
\frac{247/803}{556/803}
=
\frac{247}{556}
\approx0.4442.
$$

````r
#-- Global --#
# p
p = mean(cohorte$ever)
# no-p
not_p = 1 - mean(cohorte$ever)
# odds
odds = p/not_p

p
not_p
odds
1/odds
````

- En la proporción, comparamos fracturas con el total.
- En las odds, comparamos fracturas con los casos sin fractura.

El logit aplica el logaritmo natural a esa razón:
$$
\eta=\log\frac{p}{1-p}.
$$
Con ello, probabilidades entre 0 y 1 pasan a valores en toda la recta real. El punto central es fácil de recordar:
$$
p=0.5
\;\Longrightarrow\;
\text{odds}=1
\;\Longrightarrow\;
\eta=0.
$$

¿De dónde sale la sigmoide?
Sale de despejar $(p)$:
$$
\eta=\log\frac{p}{1-p}
\quad\Longrightarrow\quad
e^\eta=\frac{p}{1-p}
$$
$$
e^\eta(1-p)=p
\quad\Longrightarrow\quad
e^\eta=p(1+e^\eta)
$$
$$
\boxed{p=\frac{e^\eta}{1+e^\eta}}.
$$

1. Exponenciar el logit produce las odds.
2. Dividir las odds entre $(1+\text{odds})$ produce la probabilidad.

En R, `plogis()` realiza esa conversión; `qlogis()` hace la inversa.

````r
modelo_fractura <- glm(
  ever ~ x1 + x2,
  data = cohorte,
  family = binomial(link = "logit")
)
coef(modelo_fractura)

nobs(modelo_fractura)
````

La  fórmula propone:
$$
\eta_{[i]}
=
\beta_0+\beta_1x_{1,[i]}+\beta_2x_{2,[i]},
\qquad
p_{[i]}=\frac{e^{\eta_{[i]}}}{1+e^{\eta_{[i]}}}.
$$
Aquí encajan los tres componentes:
- Distribución: Bernoulli describe el resultado individual.
- Predictor: la combinación de fragilidad, tratamiento y coeficientes produce $(\eta)$.
- Enlace: el logit relaciona la media $(p)$ con ese predictor.

````r
β <- coef(modelo_fractura)

perfiles <- data.frame(x1 = c(0, 0), x2 = c(0, 1))

η_manual <- β[["(Intercept)"]] +
  β[["x1"]] * perfiles$x1 +
  β[["x2"]] * perfiles$x2

odds <- exp(η_manual)
p_manual <- odds / (1 + odds)

cbind(perfiles, η = η_manual, odds, contras=1/odds,
      fractura = p_manual, sin_fractura = 1 - p_manual)
````

Aquí también se entiende la interpretación del tratamiento:
$$
\text{odds con tratamiento}
=
e^{\beta_2}·
\text{odds sin tratamiento}.
$$

Como $(e^{\beta_2}\approx0.6068)$, el tratamiento multiplica las odds por ese factor en este modelo.

````r
α <- family(modelo_fractura)$linkinv
η <- predict(modelo_fractura, newdata = perfiles, type = "link")
cbind(perfiles, η,
      por_inversa = α(η),
      por_predict = predict(modelo_fractura, newdata = perfiles,
                            type = "response"))
````
